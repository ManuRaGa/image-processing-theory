## <font color=green> Ejercicio 1:  Importar bibliotecas necesarias</font>

```python
import cv2
import matplotlib.pyplot as plt
```

## <font color=green> Ejercicio 2: Carga y visualización de imagen a color en escala de grises</font>

**Objetivo:** Cargar y mostrar una imagen a color en escala de grises.

**Pasos:**
1.  Usa la función `cv2.imread()` con el parámetro `cv2.IMREAD_GRAYSCALE` para cargar la imagen en escala de grises. Asegurate que la imagen si sea a color.
2.   Muestra la imagen con `plt.imshow()`.

```python
img = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/1_4_Histogramas_Ecualizado/data/Escenario3.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
```

## <font color=green> Ejercicio 3:  Generar el histograma de la imagen</font>

**Objetivo:** Calcular y graficar el histograma de la imagen.

**Pasos:**
1.   Utiliza la función `cv2.calcHist()` para calcular el histograma de la imagen en escala de grises.
3.   Muestra el histograma de la imagen utilizando `plt.plot()`.

```python
hist = cv2.calcHist([img], [0], None, [256], [0,256])

plt.plot(hist, color='gray')
plt.title('Histograma de la imagen')
plt.xlabel('Intensidad de píxel')
plt.ylabel('Frecuencia')
plt.plot(hist, color='gray')
plt.xlim([0,256])

plt.show()
```

## <font color=green> Ejercicio 4:  Aplicar ecualización de histograma </font>

**Objetivo:** Mejorar el contraste de la imagen utilizando ecualización de histograma.

**Pasos:**
1.   Aplica la función `cv2.equalizeHist()` a la imagen en escala de grises.
2.   Muestra la imagen original y la imagen ecualizada, utiliza `plt.imshow()` y `plt.subplot()`, que quede una a lado de la otra.

```python
eqz = cv2.equalizeHist(img)

plt.subplot(1,2,1)
plt.title('Imagen Original')
plt.imshow(img, cmap='gray')

plt.subplot(1,2,2)
plt.title('Imagen Ecualizada')
plt.imshow(eqz, cmap='gray')

plt.show()
```

## <font color=green> Ejercicio 5: Comparar histogramas antes y después de la ecualización  </font>

**Objetivo:** Verificar los cambios que se realizaron en los datos de la imagen en relación con su grafica de histograma.

**Pasos:**
1.   Calcula el histograma de la imagen después del ecualizado utilizando `cv2.calcHist()`.
2.   Grafica ambos histogramas en subgráficos utiliza `plt.subplot()` y `plt.plot()`.



```python
hist_eqz = cv2.calcHist([eqz], [0], None, [256], [0,256])

plt.figure(figsize=(11,4))

plt.subplot(1,2,1)
plt.plot(hist)
plt.title('Histograma Original')
plt.xlabel('Intensidad de pixel')
plt.ylabel('Frecuencia')

plt.subplot(1,2,2)
plt.plot(hist_eqz)
plt.title('Histograma Ecualizado')
plt.xlabel('Intensidad de pixel')
plt.ylabel('Frecuencia')

plt.show()
```

## <font color=green> Ejercicio 6: Ecualización en imágenes a color  </font>

**Objetivo:** Aplicar ecualización de histograma a cada canal de una imagen a color.

**Pasos:**
1. Carga la imagen en color y conviértela a formato RGB.
2. Separa los canales de color (rojo, verde y azul) usando `cv2.split()`.
3. Aplica ecualización de histograma a cada canal por separado, utilizando `cv2.equalizeHist()`.
4. Une los canales y muestra la imagen original y la ecualizada, utiliza `plt.imshow()` y `plt.subplot()`, que quede una a lado de la otra.

```python
img1 = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/1_4_Histogramas_Ecualizado/data/Escenario4.jpg")
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(11,4))

r, g, b = cv2.split(img1)

r_eqz = cv2.equalizeHist(r)
g_eqz = cv2.equalizeHist(g)
b_eqz = cv2.equalizeHist(b)

img_eqz = cv2.merge([r_eqz, g_eqz, b_eqz])

plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(img1)

plt.subplot(1, 2, 2)
plt.title('Imagen Ecualizada')
plt.imshow(img_eqz)

plt.show()
```

## <font color=green> Ejercicio 7: Aplicar CLAHE (ecualización de histograma adaptativa limitada por contraste) </font>

**Objetivo:** Utilizar CLAHE para mejorar el contraste de imágenes sin aumentar demasiado el ruido.

**Pasos:**
1. Crea un objeto CLAHE `cv2.createCLAHE()` con un clip limit de 2 y tamaño de grilla 8x8.
2. Aplica CLAHE a la imagen en escala de grises `clahe.apply()`.
3. Muestra la imagen original, la imagen ecualizada y la imagen procesada con CLAHE.

```python
imagen = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
imagen2 = cv2.cvtColor(img_eqz, cv2.COLOR_RGB2GRAY)

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
img_clahe = clahe.apply(imagen)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.title('Imagen Original')
plt.imshow(imagen, cmap='gray')

plt.subplot(1,3,2)
plt.title('Imagen Ecualizada')
plt.imshow(imagen2, cmap='gray')

plt.subplot(1,3,3)
plt.title('Imagen CLAHE')
plt.imshow(img_clahe, cmap='gray')

plt.show()
```
