#  <font style="color:rgb(50,120,229)"> Dibujar con Mouse </font>

Crea una aplicación que permita lo siguiente:

- Captura de video desde una cámara web.

- Permitir al usuario dibujar un rectangulo, una linea o un circulo sobre la imagen.

- Permitir al usuario seleccionar un color de dibujo.

- Permitir al usuario salir de la aplicación.

- Todas las figuras previamente dibujadas deben ser visibles en la imagen.

EL proceso de dibujo debe ser el siguiente:

- Utilizando las teclas 'r', 'l' y 'c' el usuario selecciona el tipo de figura a dibujar.

- Utilizando tres trackbars el usuario selecciona el color de la figura a dibujar. (incluir un recuadro de color para mostrar el color seleccionado)

- Utilizando el mouse el usuario dibuja la figura seleccionada siguiendo el siguiente proceso:

    - Presionar el botón izquierdo del mouse y sin soltarlo arrastrar el mouse hasta el punto donde se desea terminar la figura.

    - Soltar el botón izquierdo del mouse.

    - Repetir el proceso para dibujar otra figura.

    - Al mover el mouse sin presionar ningún botón se debe mostrar la figura que se está dibujando en tiempo real.

- El usuario puede salir de la aplicación presionando la tecla 'q'.
