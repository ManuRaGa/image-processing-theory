#  <font style="color:rgb(50,120,229)"> Cambio de anuncio publicitario</font>

Crea una aplicación que permita lo siguiente:

* Cargar una imagen de fondo con pantallas publicitarias.
* Cargar una imagen secundaria que se desee insertar como anuncio.
* Permitir al usuario seleccionar cuatro puntos con el mouse sobre la imagen de fondo, que representen los vértices de la superficie donde se desea colocar el anuncio.
* Realizar la transformación de perspectiva (homografía) para deformar la imagen del anuncio y que se adapte a la superficie seleccionada.
* Insertar el anuncio sobre la imagen original, respetando la perspectiva y sin dañar el fondo.

El proceso de interacción con el usuario debe ser el siguiente:
* Se muestra la imagen de fondo en una ventana.
* El usuario debe hacer clic en cuatro puntos de la imagen, en sentido horario o antihorario, que correspondan al área donde desea insertar el anuncio.
* Cada vez que se haga clic, se debe dibujar un círculo rojo en el punto seleccionado para confirmar la elección.
* Una vez seleccionados los cuatro puntos, se debe cerrar la ventana automáticamente o tras presionar una tecla.
* La aplicación realiza la deformación de la imagen del anuncio (usando homografía) para que encaje con la región seleccionada, mostrar la imagen de como queda ajustada.
* Finalmente, se inserta la imagen transformada dentro del área seleccionada, enmascarando correctamente el área original.
* La imagen resultante se muestra en una nueva ventana nueva.
