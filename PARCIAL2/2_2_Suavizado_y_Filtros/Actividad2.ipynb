## <font color=green> Ejercicio 2: Filtro Gaussiano para Reducción de Ruido</font>

**Objetivo:** Aplicar un filtro gaussiano para suavizar una imagen y reducir el ruido gaussiano.

**Pasos:**
1.   Cargar la imagen original en escala de grises usando `cv2.imread().`
2.   Cargar la imagen ruido_g en escala de grises usando `cv2.imread().`
3.   Aplicar un filtro gaussiano con `cv2.GaussianBlur()`.
4.   Visualizar la imagen original, la imagen con ruido y la imagen filtrada usando `plt.subplot().`

```python
original = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/PARCIAL2/2_2_Suavizado_y_Filtros/data/Original.jpg", cv2.IMREAD_GRAYSCALE)
ruidog = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/PARCIAL2/2_2_Suavizado_y_Filtros/data/ruido_g.jpg", cv2.IMREAD_GRAYSCALE)

f_gauss = cv2.GaussianBlur(ruidog, (21,21), 10.0)

plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(original, cmap='gray')
plt.title('Imagen Original')

plt.subplot(1,3,2)
plt.imshow(ruidog, cmap='gray')
plt.title('Imagen con Ruido Gaussiano')

plt.subplot(1,3,3)
plt.imshow(f_gauss, cmap='gray')
plt.title('Imagen con Filtro Gausiano')
plt.show()
```

## <font color=green> Ejercicio 3: Filtro simple para Ruido Uniforme </font>

**Objetivo:** Aplicar un filtro simple para reducir el ruido uniforme en una imagen.

**Pasos:**
1.   Cargar la imagen ruido_u en escala de grises usando `cv2.imread().`
2.   Aplicar un filtro de media con `cv2.blur()`.
3.   Visualizar la imagen original, la imagen con ruido y la imagen filtrada usando `plt.subplot().`

```python
ruidou = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/PARCIAL2/2_2_Suavizado_y_Filtros/data/ruido_u.jpg", cv2.IMREAD_GRAYSCALE)

f_blurr = cv2.blur(ruidou, (16,16))

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(ruidou, cmap='gray')
plt.title('Imagen con Ruido Uniforme')

plt.subplot(1,2,2)
plt.imshow(f_blurr, cmap='gray')
plt.title('Imagen con Filtro Simple')
```

## <font color=green> Ejercicio 4:  Filtro Mediana para Ruido Sal y Pimienta </font>

**Objetivo:** Aplicar un filtro mediana para eliminar el ruido sal y pimienta.

**Pasos:**
1.   Cargar la imagen ruido_s_p en escala de grises usando `cv2.imread().`
2.   Aplicar un filtro mediana con `cv2.medianBlur()`.
3.   Visualizar la imagen original, la imagen con ruido y la imagen filtrada usando `plt.subplot().`

```python
ruidosp = cv2.imread("C:/Users/HUAWEI/Desktop/TRATAMIENTO/PARCIAL2/2_2_Suavizado_y_Filtros/data/ruido_s_p.jpg", cv2.IMREAD_GRAYSCALE)

f_mblurr = cv2.blur(ruidosp, (16,16))

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(ruidosp, cmap='gray')
plt.title('Imagen con Ruido Sal Pimienta')

plt.subplot(1,2,2)
plt.imshow(f_mblurr, cmap='gray')
plt.title('Imagen con Filtro Mediana')
```

```python
bilateral1 = cv2.bilateralFilter(f_gauss, 9, 75, 75)
bilateral2 = cv2.bilateralFilter(f_blurr, 9, 75, 75)
bilateral3 = cv2.bilateralFilter(f_mblurr, 9, 75, 75)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(bilateral1, cmap='gray')
plt.title('Imagen con Filtro Bilateral en Gauss')

plt.subplot(1,2,2)
plt.imshow(bilateral2, cmap='gray')
plt.title('Imagen con Filtro Bilateral en Simple')

plt.subplot(1,3,2)
plt.imshow(bilateral3, cmap='gray')
plt.title('Imagen con Filtro Bilateral en Media')
```
