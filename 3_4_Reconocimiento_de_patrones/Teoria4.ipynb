# <font color=blue>Reconocimiento de Patrones</font>


## <font color="#F20C60">¿Qué es el reconocimiento de patrones?</font>

El reconocimiento de patrones es una disciplina encargada de identificar figuras, reconocer formas o leer patrones, con la finalidad de recopilar información sobre el objeto que se está estudiando y asignarlo a un grupo o clase.

Para entender este concepto de manera más fácil, se tienen los siguientes términos:

 * **Reconocer:** Distinguir un objeto entre varios, ya que dicho objeto posee características y atributos propios que lo diferencia del resto.
 * **Patrón:** Es un objeto que presenta cierto comportamiento y/o características ya conocidas.
 * **Clase:** Es un grupo de objetos que comparten los mismos atributos y/o comportamientos.
 * **Atributo:** Es cualquier medida o cualidad extraíble. Constituye los parámetros, factores, variables y comportamientos.

En el ámbito de la inteligencia artificial (IA), el reconocimiento de patrones es un componente vital. Los sistemas de IA pueden utilizar los patrones identificados para tomar decisiones o hacer predicciones.


## <font color="#F20C60">¿Cómo funciona el reconocimiento de patrones?</font>

Se tienen varios pasos los cuales son:

1. **Adquisición de datos:** Este puede ser apartir de camaras, sensores o cualquier otro dispositivo que transforme magnitudes físicas o químicas en magnitudes eléctricas. Las variables captadas pueden ser distintas como, por ejemplo: color, temperatura, intensidad lumínica, inclinación, velocidad, fuerza, etc.
2. **Formulación de característica:** Consta en la formulación de características para usarse posteriormente en la clasificación de objetos.
3. **Selección de atributos:** Es la selección adecuada de atributos para describir a los objetos.
4. **Clasificación de objetos:** En este paso, se clasifican los objetos de acuerdo a los atributos que presentan, asignándolos en un grupo u otro. En este procedimiento se utilizan tecnologías de machine learning, que consta en hacer que las computadoras aprendan a clasificar los objetos de manera autónoma.
  * **Clasificación supervisada:** Es la utilización de modelos ya prestablecidos para clasificar objetos o comportamientos.
  * **Clasificación no supervisada:** Es cuando la clasificación se realiza intentando identificar similitudes entre objetos.
  * **Clasificación parcialmente supervisada:** Es un panorama donde existen modelos en algunas clases, pero no en todas. Por lo tanto, combina elementos de la clasificación supervisada y no supervisada.

## <font color="#F20C60">Aplicaciones del reconocimiento de patrones</font>


* **Reconocimiento facial:** Es un sistema capaz de reconocer automáticamente un rostro en un espacio dado.
* **Reconocimiento de voz:** Es la capacidad que posee un sistema inteligente en reconocer palabras y oraciones en un idioma determinado.
* **Reconocimiento de huella dactilar:** Son dispositivos que identifican huellas dactilares, y detectar si dicha huella está almacenada en alguna base de datos.
* **Reconocimiento de caracteres:** Es un dispositivo especializado en reconocer caracteres, tanto escritos a mano, como escritos digitalmente. Estos caracteres son presentados al dispositivo como una imagen, y el sistema procede a almacenar estos caracteres, poniéndolos a disposición del usuario para que pueda usarlos posteriormente en un programa de edición de texto.
* **Reconocimiento de objetos:** Es usado en conjunto con la visión artificial. Y consta en la detección de objetos en una imagen o secuencia de imágenes (vídeo).
* **Reconocimiento de música:** Identifica qué música está sonando.
* **Reconocimiento de sonidos:** Distingue qué sonido está produciéndose. Como, por ejemplo, los ladridos de un perro, un bebé llorando, campanada de un timbre, etc.
* **Reconocimiento de anomalías en imágenes de ratos X:** Identifica si existen irregularidades en proyecciones de rayos X, con el fin de determinar un problema existente en personas o animales.
* **Reconocimiento de células infectadas:** Es capaz de detectar células con partículas víricas en su interior.
* **Reconocimiento de situaciones meteorológicas:** Permite pronosticar eventos meteorológicos de acuerdo a diversos patrones.

## <font color="#F20C60">Perímetro del contorno</font>

También se denomina longitud de arco, y se utiliza principalmente para obtener el valor del perimetro de un contorno.

### <font color=green> Sintaxis de la función </font>
``` python
p = cv.arcLength (cnt, True )
```

* `cnt:` Contornos de la imagen.
* `flag:` True indica que el contorno está cerrado (forma cerrada como un círculo o rectángulo). Si fuera una curva abierta, usarías False.
* `p:` resultado del perimetro.

### <font color=orange> Ejemplo </font>

Creamos una imagen con una resolución de 200 x 200 y dibujamos un rectangulo blanco.

````python
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (150, 100), 255, -1)
````

Buscamos los contornos en la imagen

````python
contours, _ = cv2.findContours(img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

print(f"Se encontraron {len(contours)} contornos")
````

Calculamos el perimetro

````python
for i, cnt in enumerate(contours):
    perimetro = cv2.arcLength(cnt, True)
    print(f"Perímetro del contorno {i + 1}: {perimetro}")
````

## <font color="#F20C60">Aproximación de contornos</font>

Aproxima una forma de contorno a otra con menos vértices, dependiendo de la precisión especificada. Es una implementación del algoritmo de [Douglas-Peucker](https://en.wikipedia.org/wiki/Ramer%E2%80%93Douglas%E2%80%93Peucker_algorithm).

Para entender esto, supongamos que intentas encontrar un cuadrado en una imagen, pero debido a algunos problemas, no obtienes un cuadrado perfecto, sino una forma incorrecta (como se muestra en la primera imagen a continuación). Ahora puedes usar esta función para aproximar la forma. En este caso, el segundo argumento se llama épsilon y representa la distancia máxima entre el contorno y el contorno aproximado. Es un parámetro de precisión. Se requiere una selección adecuada de épsilon para obtener el resultado correcto. En la segunda imagen, la línea verde muestra la curva aproximada para épsilon = 10 % de la longitud del arco. La tercera imagen muestra lo mismo para épsilon = 1 % de la longitud del arco.

image.png


### <font color=green> Sintaxis de la función </font>

``` python
aprox = cv.approxPolyDP (cnt,epsilon,flag)
```

* `cnt`: Contornos de la imagen.
* `epsilon`: representa la distancia máxima entre el contorno y el contorno aproximado.
* `flag`: True indica que el contorno está cerrado (forma cerrada como un círculo o rectángulo). Si fuera una curva abierta, usarías False.
* `aprox`: Resultado de la aproximación.


### <font color=orange> Ejemplo </font>

Se aplica la aproximación a los perimetros encontrados apartir de los contornos

````python
for i, cnt in enumerate(contours):
    aprox = cv2.approxPolyDP (cnt,perimetro,True)
    print(f"Perímetro del contorno {i + 1}: {aprox}")
````

## <font color="#F20C60">Envoltura convexa</font>

La envoltura convexa se parece a la aproximación de contornos, pero no lo es (ambas pueden ofrecer los mismos resultados en algunos casos). En este caso, la función `cv.convexHull()` comprueba si una curva tiene defectos de convexidad y los corrige. En general, las curvas convexas son aquellas que siempre presentan una curvatura abombada o, al menos, plana. Si presentan una curvatura abombada en el interior, se denominan defectos de convexidad. Por ejemplo, observe la imagen de una mano a continuación. La línea roja muestra la envoltura convexa de la mano. Las flechas de doble punta indican los defectos de convexidad, que son las desviaciones locales máximas de la envoltura con respecto a los contornos. Para mas información da clic [aqui](https://docs.opencv.org/4.x/dd/d49/tutorial_py_contour_features.html).

image.png

### <font color=green> Sintaxis de la función </font>

``` python
out = cv.convexHull(points[, hull[, clockwise[, returnPoints]]])
```

* `points:` son los contornos por los que pasamos.
* `hull:` es la salida, normalmente lo evitamos.
* `clockwise:` Indicador de orientación. Si es verdadero, la envolvente convexa de salida se orienta en sentido horario. De lo contrario, se orienta en sentido antihorario.
* `returnPoints:` Por defecto, True. Devuelve las coordenadas de los puntos de la envolvente. Si es False, devuelve los índices de los puntos de contorno correspondientes a los puntos de la envolvente.

## <font color="#F20C60">Rectángulo delimitador</font>

Hay dos tipos de rectángulos delimitadores.
* **Rectángulo delimitador recto**: no considera la rotación del objeto. Por lo tanto, el área del rectángulo delimitador no será mínima. Se calcula mediante la función `cv.boundingRect()`.
* **Rectángulo rotado**:Aquí, el rectángulo delimitador se dibuja con un área mínima, por lo que también considera la rotación. La función utilizada es `cv.minAreaRect()` . Para dibujar este rectángulo, necesitamos cuatro esquinas. Estas se obtienen mediante la función `cv.boxPoints()`.

### <font color=green> Sintaxis de la función Recto </font>

``` python
x,y,w,h = cv.boundingRect(cnt)
```

Sea `(x,y)` la coordenada superior izquierda del rectángulo y `(w,h)` su ancho y alto.

### <font color=green> Sintaxis de la función Rotador </font>

``` python
rect = cv.minAreaRect(cnt)
box = cv.boxPoints(rect)
```

* `cnt`: son los contornos.
* `rect`: estructura que contiene los valores del centro, ancho, alto y ámgulo de rotación.
* `box`: cuatro esquinas de la estructura para poder dibujar.

## <font color="#F20C60">Colocar texto</font>

Este método se utiliza para dibujar una cadena de texto en cualquier imagen.

### <font color=green> Sintaxis de la función</font>

``` python
cv2.putText(image, text, org, font, fontScale, color[, thickness[, lineType[, bottomLeftOrigin]]])  
```

* `image`: Es la imagen en la que se dibujará el texto.
* `text` : Cadena de texto que se dibujará.
* `org`: Son las coordenadas de la esquina inferior izquierda de la cadena de texto en la imagen. Las coordenadas se representan como tuplas de dos valores, es decir ( valor de la coordenada X , valor de la coordenada Y ) .
* `font`: Indica el tipo de fuente. Algunos tipos de fuente son FONT_HERSHEY_SIMPLEX, FONT_HERSHEY_PLAIN, etc.
* `fontScale` : Factor de escala de fuente que se multiplica por el tamaño base específico de la fuente.
* `color` : Es el color de la cadena de texto que se dibujará.
* `thurst` : Es el grosor de la línea en px .
* `lineType`: Este parámetro es opcional. Indica el tipo de línea que se utilizará. bottomLeftOrigin: Este parámetro es opcional. Cuando es verdadero, el origen de los datos de la imagen se encuentra en la esquina inferior izquierda. De lo contrario, se encuentra en la esquina superior izquierda.

## <font color="#F20C60">Dibujar rectangulo</font>

Este método se utiliza para dibujar un rectángulo en cualquier imagen

### <font color=green> Sintaxis de la función</font>

``` python
cv2.rectangle(image, start_point, end_point, color, thickness)  
```

* `image`: Es la imagen en la que se dibujará el rectángulo.
* `start_point`: Son las coordenadas iniciales del rectángulo. Las coordenadas se representan como tuplas de dos valores, es decir ( valor de la coordenada X , valor de la coordenada Y ).
* `end_point`: Son las coordenadas finales del rectángulo. Las coordenadas se representan como tuplas de dos valores, es decir ( valor de la coordenada X , valor de la coordenada Y ).
* `color`: Es el color de la línea del borde del rectángulo que se dibujará
* `thickness`: Es el grosor de la línea del borde del rectángulo en px . Un grosor de -1 px rellenará la forma del rectángulo con el color especificado.
