# <font color=blue>Detección de caracteristicas y esquinas</font>


Este notebook tiene como objetivo proporcionar las bases necesarias para poder trabajar con la detección de características y algoritmos como Shi-Tomasi, Harris, SIFT, SURF y ORB, así como el concepto de matching de características

## <font color="#F20C60">¿Qué es una característica en una imagen?</font>

Son puntos o patrones distintivos en una imagen, como esquinas, bordes o texturas, que pueden ser detectadas y comparadas entre diferentes imágenes.

## <font color="#F20C60">¿Qué son las esquinas y por qué son importantes?</font>

Son puntos donde dos bordes se cruzan y la intensidad de la imagen cambia en varias direcciones.Son útiles porque son fáciles de localizar y comparar en diferentes imágenes.

## <font color="#F20C60">Algoritmo Harris</font>

El Detector de esquinas de Harris es uno de los algoritmos más antiguos y conocidos para detectar esquinas en imágenes. Analiza la variación de intensidad de los píxeles en pequeñas regiones, y detecta aquellos puntos donde la intensidad cambia en múltiples direcciones, lo cual corresponde a una esquina.

### <font color="#770EB2">Características principales:</font>

* Detecta esquinas de manera eficiente.
* Robusto ante ruido, pero sensible a cambios de escala.
* No detecta descriptores, solo las posiciones de las esquinas.

### <font color=green> Sintaxis de la función </font>
``` python
cv2.cornerHarris(src, blockSize, ksize, k)
```

* `src`: Imagen de entrada (en escala de grises, tipo float32).
* `blockSize`: Tamaño del vecindario considerado para la detección de esquinas (ej. 2, 3).
* `ksize`: Tamaño del kernel de la derivada de Sobel (ej. 3, 5).
* `k`: Parámetro libre del detector de Harris (generalmente entre 0.04 y 0.06).


## <font color="#F20C60">Algoritmo Shi-Tomasi</font>


El algoritmo Shi-Tomasi es una mejora del detector de Harris. También busca esquinas, pero utiliza un criterio diferente para elegir las mejores. En general, Shi-Tomasi es más preciso y estable al identificar las esquinas importantes.

### <font color="#770EB2">Características principales:</font>

* Más preciso que Harris para esquinas.
* Selecciona solo las "mejores" esquinas según una puntuación.
* Ideal para seguimiento de esquinas en video.

### <font color=green> Sintaxis de la función </font>
``` python
cv2.goodFeaturesToTrack(image, maxCorners, qualityLevel, minDistance)
```



* `image`: Imagen de entrada en escala de grises.
* `maxCorners`: Número máximo de esquinas a devolver.
* `qualityLevel`: Valor mínimo de calidad (entre 0 y 1). Cuanto más alto, menos esquinas se detectan.
* `minDistance`: Distancia mínima entre las esquinas detectadas.

## <font color="#F20C60">Algoritmo SIFT (Scale-Invariant Feature Transform)</font>

SIFT es un algoritmo que detecta y describe características locales en imágenes. A diferencia de Harris y Shi-Tomasi, SIFT no solo detecta puntos clave, sino que también calcula descriptores que permiten comparar características entre imágenes.

### <font color="#770EB2">Características principales:</font>

* Invariante a escala (detecta características sin importar el tamaño).
* Invariante a rotación (funciona aunque la imagen esté girada).
* Genera descriptores únicos para cada punto detectado.

### <font color=green> Sintaxis de la función </font>
``` python
sift = cv2.SIFT_create()
keypoints, descriptors = sift.detectAndCompute(image, None)
```

* `image`: Imagen en escala de grises.
* `None`: Máscara (puede usarse para indicar zonas específicas a analizar, si no se usa se pone None).
* `keypoints`: Lista de puntos clave detectados.
* `descriptors`: Descriptores de cada punto (vectores de características).

## <font color="#F20C60">Algoritmo SURF (Speeded-Up Robust Features)</font>

SURF es un algoritmo basado en SIFT, pero diseñado para ser más rápido y eficiente. También detecta puntos clave y genera descriptores. Aunque ya no es libre en todas sus versiones, es importante conocerlo por sus aplicaciones en sistemas donde la velocidad es crítica.

### <font color="#770EB2">Características principales:</font>

* Más rápido que SIFT.
* También invariante a escala y rotación.
* Utiliza aproximaciones para acelerar el cálculo.

**Nota:** SURF requiere versiones específicas de OpenCV con contrib modules debido a temas de licencia.

### <font color=green> Sintaxis de la función </font>
``` python
surf = cv2.xfeatures2d.SURF_create()
keypoints, descriptors = surf.detectAndCompute(image, None)
```

* `image`: Imagen en escala de grises.
* `None`: Máscara (como en SIFT).
* `keypoints`: Puntos clave
* `descriptors`: Descriptores generados.

## <font color="#F20C60">Algoritmo ORB (Oriented FAST and Rotated BRIEF)</font>

ORB es una alternativa gratuita y rápida a SIFT y SURF. Combina el detector FAST para encontrar puntos clave y el descriptor BRIEF para describirlos. Además, ORB introduce mejoras para soportar rotaciones y pequeñas variaciones de escala.

### <font color="#770EB2">Características principales:</font>

* Muy rápido.
* Gratuito y disponible en OpenCV.
* Funciona bien para aplicaciones en tiempo real.



### <font color=green> Sintaxis de la función </font>
``` python
orb = cv2.ORB_create()
keypoints, descriptors = orb.detectAndCompute(image, None)
```

* `image`: Imagen en escala de grises.
* `None`: Máscara opcional.
* `keypoints`: Puntos clave detectados.
* `descriptors`: Descriptores generados.



## <font color="#F20C60">Matching de Imágenes (Emparejamiento de Características)</font>

El matching o emparejamiento de características consiste en comparar puntos clave y sus descriptores entre dos imágenes para encontrar correspondencias. Es fundamental para tareas como reconocimiento de objetos, seguimiento y ensamblaje de panoramas.

### <font color="#770EB2">Proceso básico:</font>

1. Detectar características (puntos clave) en ambas imágenes.
2. Generar descriptores para cada característica.
3. Comparar los descriptores para encontrar las mejores coincidencias.

### <font color=green> Sintaxis de la función </font>
``` python
bf = cv2.BFMatcher()
matches = bf.match(des1, des2)
```

* `des1`: Descriptores de la primera imagen.
* `des2`: Descriptores de la segunda imagen.
* `matches`: Lista de correspondencias encontradas.

### <font color="#770EB2">KNN Matching</font>

### <font color=green> Sintaxis de la función </font>
``` python
matches = bf.knnMatch(des1, des2, k=2)
```

* `k=2`: Devuelve los dos mejores matches por cada descriptor (para aplicar técnicas como Lowe's ratio test).
